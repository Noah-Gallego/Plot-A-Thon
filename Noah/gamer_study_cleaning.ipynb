```python
# Import Libraries
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Load in Data
url = 'https://raw.githubusercontent.com/Noah-Gallego/Plot-A-Thon/main/gamer_study_raw.csv'
df = pd.read_csv(url)
```

```python
# Set Index to ID
df = df.set_index('respondent_id')
df.head()
```

# Handle Erroneous Values and Clean! 🧹

```python
# Deal with Missing Values
df['age'].unique()
```

We have an age of -3! As well as some missing values. Let's deal with those.

```python
# Replace erroneous value
df = df.replace(-3, np.nan)
```

```python
# Display Unique Values of all Numerical Columns
for col in df.columns:
  if df[col].dtype == 'int64' or df[col].dtype == 'float64':
    print(col, df[col].unique())
```

All other numerical columns looks good! Let's inspect the categorical ones!

```python
for col in df.columns:
  if df[col].dtype == 'object':
    print(col, df[col].unique())
```

```python
df.head()
```

It appears there are several erroneous values, but intent is known. For example 'MALE ' should clearly be mapped to 'Male' and 'femal' to 'Female'. So let's clean that!

```python
# Clean Gender Column by mapping 'MALE ' to 'Male' and 'femal' to 'Female'
df['gender'] = df['gender'].replace({'MALE ': 'Male', 'femal': 'Female'})

# Clean Race/Ethnicity Column by Mapping 'Whtie' to 'White/Caucasian'
df['race_ethnicity_primary'] = df['race_ethnicity_primary'].replace({'Whtie': 'White/Caucasian'})

# Clean is_player by mapping 'No' to 0 and 'yes' to 1
df['is_player'] = df['is_player'].map({'No': 0, 'yes': 1, 'FALSE': 0, 'TRUE': 1}).astype('Int64')
```

```python
# Print all Column Data Types
print(list(df.dtypes))
```

```python
# Loop over each column and convert booleans to 0:1
for col in df.columns:
  if df[col].dtype == 'bool':
    df[col] = df[col].astype('Int64')

# Convert LGBT Column
df['lgbt_identification'] = df['lgbt_identification'].map({'No': 0, 'Yes': 1})
```
